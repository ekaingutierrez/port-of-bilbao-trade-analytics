```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

data_macro = {
    'Year': [2020, 2021, 2022, 2023, 2024, 2025],
    'GDP Growth (%)': [-3.9, 6.2, 4.3, 0.1, 1.4, 1.8],
    'Unemployment (%)': [4.1, 4.2, 3.5, 3.6, 3.7, 3.8],
    'Inflation (%)': [1.3, 2.7, 10.0, 3.8, 2.8, 2.4],
    'House Price Index (2015=100)': [145, 168, 192, 185, 194, 202],
    'Gov Deficit (% GDP)': [0.5, -2.2, -0.1, -0.3, -1.8, -2.1], # Balance Fiscal
    'Trade Surplus (Bn EUR)': [72.4, 78.1, 85.3, 91.2, 94.5, 98.2],
    'Population (Millions)': [17.4, 17.5, 17.7, 17.8, 17.9, 18.0],
    'Renewable Share (%)': [13.2, 14.1, 15.0, 16.5, 18.2, 20.1]
}

df = pd.DataFrame(data_macro)
```

```python
plt.figure(figsize=(8,5))
plt.plot(df['Year'], df['GDP Growth (%)'], marker='o', color='skyblue', lw=2)
plt.title('Crecimiento Económico: PIB Real (%)')
plt.xlabel('Año')
plt.ylabel('Crecimiento (%)')
plt.axhline(0, color='black', lw=1)
plt.grid(True, alpha=0.3)
plt.show()
```

```python
plt.figure(figsize=(8,5))
plt.plot(df['Year'], df['Unemployment (%)'], marker='s', color='darkblue', lw=2)
plt.title('Tasa de Desempleo (Netherlands)')
plt.xlabel('Año')
plt.ylabel('% fuerza laboral')
plt.ylim(0,6)
plt.grid(True, alpha=0.3)
plt.show()
```

```python
fig, ax1 = plt.subplots(figsize=(8,5))
ax1.plot(df['Year'], df['Inflation (%)'], marker='d', color='red', label='Inflación')
ax1.set_xlabel('Año')
ax1.set_ylabel('Inflación (%)', color='red')
ax1.grid(True, alpha=0.3)
ax2 = ax1.twinx()
ax2.plot(df['Year'], df['House Price Index (2015=100)'], linestyle='--', color='lightblue', label='HPI')
ax2.set_ylabel('Índice Vivienda', color='lightblue')
plt.title('Inflación y Precio Vivienda (Evolución)')
plt.show()
```

```python
plt.figure(figsize=(8,5))
plt.bar(df['Year'], df['Gov Deficit (% GDP)'], color='skyblue', alpha=0.7)
plt.title('Balance Fiscal (% PIB)')
plt.xlabel('Año')
plt.ylabel('Déficit / Superávit')
plt.axhline(0, color='black', lw=0.8)
plt.grid(True, axis='y', alpha=0.3)
plt.show()
```

```python
plt.figure(figsize=(8,5))
plt.fill_between(df['Year'], df['Trade Surplus (Bn EUR)'], color='green', alpha=0.2)
plt.plot(df['Year'], df['Trade Surplus (Bn EUR)'], color='green', lw=2)
plt.title('Capacidad Exportadora: Superávit Comercial')
plt.xlabel('Año')
plt.ylabel('Miles de millones €')
plt.grid(True, alpha=0.3)
plt.show()
```

```python
plt.figure(figsize=(8,5))
plt.plot(df['Year'], df['Population (Millions)'], marker='o', color='purple', lw=2)
plt.title('Evolución Demográfica: Población Total')
plt.xlabel('Año')
plt.ylabel('Millones de habitantes')
plt.grid(True, alpha=0.3)
plt.show()
```

```python
plt.figure(figsize=(8,5))
plt.bar(df['Year'], df['Renewable Share (%)'], color='lightgreen', alpha=0.7)
plt.title('Transición Energética: % Energía Renovable')
plt.xlabel('Año')
plt.ylabel('% del total')
plt.grid(True, axis='y', alpha=0.3)
plt.show()
```

```python
plt.figure(figsize=(8,5))
indicators = ['Crecimiento', 'Desempleo', 'Inflación', 'Déficit']
values = [
    df.iloc[-2]['GDP Growth (%)'],
    df.iloc[-2]['Unemployment (%)'],
    df.iloc[-2]['Inflation (%)'],
    df.iloc[-2]['Gov Deficit (% GDP)']
]
plt.barh(indicators, values, color=['skyblue', 'darkblue', 'red', 'orange'])
plt.title('Resumen Ejecutivo: Indicadores Clave 2024')
plt.xlabel('Valor (%)')
plt.axvline(0, color='black', lw=1)
plt.grid(True, axis='x', alpha=0.3)
plt.show()
```
